### Word Document Processing 

```python
from langchain_community.document_loaders import Docx2txtLoader, UnstructuredWordDocumentLoader
```

```python
# Method 1: Using DocxtxtLoader
print("1. Using Docx2txtLoader")

try:
    docx_loader = Docx2txtLoader("data/word_files/RAG_Notes_practice.docx")
    docs = docx_loader.load()
    print(f"Loaded {len(docs)} document(s)")
    print(f"\nContent preview: {docs[0].page_content[:200]}...")
    print(f"Metadata: {docs[0].metadata}")

except Exception as e:
    print(f"Error: {e}")
```

```python
# Method 2: Using UnstructuredWordDocumentLoader
print("2. UnstructuredWordDocumentLoader")

try:
    unstructured_loader = UnstructuredWordDocumentLoader("data/word_files/RAG_Notes_practice.docx", mode = "elements")    
    # elements mode is used to preserve the structure of the document
    unstructured_docs = unstructured_loader.load() 

    print(f"Loaded {len(unstructured_docs)} elements")
    for i, doc in enumerate(unstructured_docs[:3]):
        print(f"\nElements {i+1}:")
        print(f"Type: {doc.metadata.get('category', 'unknown')}")
        print(f"Content: {doc.page_content[:100]}...")

except Exception as e:
    print(e)
```
